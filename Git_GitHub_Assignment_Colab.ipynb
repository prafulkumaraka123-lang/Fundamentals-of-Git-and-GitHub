## Question 1 – What is Git, and how is it different from GitHub? Explain distributed and centralized version control.

**Answer:** Git is a free, open-source **distributed version control system (DVCS)**. It tracks file changes, records project history, and lets developers compare, restore, and collaborate on different versions.

**GitHub** is an online hosting and collaboration platform for Git repositories. It provides remote repository hosting and tools such as Pull Requests, code reviews, issue tracking, and project collaboration.

- **Git:** Version-control software used to manage project history, usually on a developer's computer.
- **GitHub:** A web-based platform for hosting Git repositories and collaborating.

**Distributed version control:** Each developer has a complete local repository copy, including its history. Commits can be made locally without an internet connection and shared later.

**Centralized version control:** A central server stores the main version history. Developers depend more heavily on that server to obtain versions and share changes. If the central server is unavailable, many version-control operations may be blocked.

**Main difference:** A Git clone contains repository history locally, while a centralized system relies on a central server as the primary location for version history.

## Question 2 – Difference between `git init` and `git clone`. Give a situation for each.

**Answer:**

- **`git init`** creates a new, empty Git repository in the current directory by initializing the `.git` directory. Use it when starting a new local project that is not already under version control.
- **`git clone`** creates a local copy of an existing remote repository, including its files and history, and usually configures a remote named `origin`. Use it when joining or contributing to an existing GitHub project.

Examples:

```bash
# Start version control for a new project
mkdir my-project
cd my-project
git init

# Copy an existing remote repository
git clone https://github.com/USERNAME/REPOSITORY.git
```

## Question 3 – Explain the Working Directory, Staging Area, and Repository. Give the commands for `notes.txt`.

**Answer:**

1. **Working Directory:** The project files that a developer views and edits. Changes here are not yet included in a commit.
2. **Staging Area (Index):** The area where selected changes are prepared for the next commit. `git add` stages changes.
3. **Repository:** The Git database that stores committed snapshots and project history. `git commit` saves staged changes.

Command sequence:

```bash
# Create or edit the file in the working directory
echo "My project notes" > notes.txt

# Check the working tree
git status

# Move changes to the staging area
git add notes.txt

# Save the staged change in the repository
git commit -m "Add notes file"
```

The sequence is **Working Directory → Staging Area → Repository**.

## Question 4 – What is a branch, why is branching useful, and how do you create and switch to `feature-payment`?

**Answer:** A **branch** is a separate line of development in a Git repository. It allows developers to work on features or bug fixes without immediately changing the main development branch.

Branching is useful because:
- Developers can work on separate tasks in parallel.
- New work can be developed and tested without disturbing the main branch.
- Changes can be reviewed before being merged.
- Different features can progress independently.

Commands:

```bash
git branch feature-payment
git switch feature-payment
```

A shorter modern command creates and switches to the branch in one step:

```bash

```

The older equivalent is `git checkout -b feature-payment`.

## Question 5 – Difference between `git merge` and `git rebase`. Give one advantage and one risk of rebase on a shared branch.

**Answer:**

- **`git merge`** combines changes from another branch into the current branch. It preserves the original branch history and may create a merge commit.
- **`git rebase`** replays commits from the current branch on top of another base commit. It creates new commit IDs for the replayed commits and can produce a linear history.

**Advantage of rebase:** It can make project history cleaner and more linear, making the sequence of changes easier to follow.

**Risk on a shared branch:** Rebase rewrites history by creating new commit IDs. Other developers may already have based work on the old commits, so rewriting them can cause divergent histories and require coordination.

**Good practice:** Avoid rebasing commits that have already been shared unless the team has coordinated it. Merging is generally safer for integrating shared branch history.

## Question 6 – Two developers modify the same line on different branches. What happens during merge, and how do you resolve it?

**Answer:** If two developers change the same line in different ways, Git may not be able to determine which version should be kept. During a merge, Git reports a **merge conflict** and pauses until it is resolved.

**Resolution steps:**

1. Switch to the branch that should receive the changes:
   ```bash
   git switch main
   ```
2. Merge the other branch:
   ```bash
   git merge feature-branch
   ```
3. If Git reports a conflict, identify the files:
   ```bash
   git status
   ```
4. Open each conflicted file. Git marks the conflicting sections, for example:
   ```text
   <<<<<<< HEAD
   Content from the current branch
   =======
   Content from the branch being merged
   >>>>>>> feature-branch
   ```
5. Decide which content should remain, or combine both changes. Remove the conflict markers and save the file.
6. Stage the resolved file:
   ```bash
   git add filename
   ```
7. Complete the merge:
   ```bash
   git commit
   ```
8. Verify the result:
   ```bash
   git status
   git log --oneline --graph --all
   ```

Review the final file and run relevant tests. Do not choose one side without checking the intended changes.

## Question 8 – Difference between `git fetch` and `git pull`. Write commands to fetch and merge manually.

**Answer:**

- **`git fetch`** downloads new commits and updated references from a remote repository, updating remote-tracking references such as `origin/main`. It does not automatically merge the changes into the current local branch.
- **`git pull`** fetches remote changes and then integrates them into the current branch. It is commonly a fetch followed by a merge, unless configured to rebase.

Fetch and merge manually:

```bash
# Download updates from origin
git fetch origin

# Inspect status and incoming history
git status
git log --oneline --graph --decorate --all

# Merge the remote main branch into the current branch
git merge origin/main
```

Use the correct remote branch name for your project. For example, replace `main` if the remote branch has another name.

## Question 9 – What is a Pull Request on GitHub? Describe the typical workflow.

**Answer:** A **Pull Request (PR)** is a GitHub feature used to propose that changes from one branch be reviewed and merged into another branch. It allows team members to discuss changes, review code, request improvements, and run automated checks before integration.

Pull Requests are important because they support code review, communication, quality checks, bug detection, and coordinated collaboration.

**Typical workflow:**

1. Clone the repository:
   ```bash
   git clone https://github.com/USERNAME/REPOSITORY.git
   cd REPOSITORY
   ```
2. Create a feature branch:
   ```bash
   git switch -c feature-payment
   ```
3. Make the required code changes.
4. Stage and commit:
   ```bash
   git add .
   git commit -m "Add payment feature"
   ```
5. Push the branch:
   ```bash
   git push -u origin feature-payment
   ```
6. On GitHub, create a Pull Request with `main` as the base branch and `feature-payment` as the compare branch.
7. Review and discuss the changes. Make requested updates, commit them, and push to the same feature branch.
8. Merge the Pull Request after approval and required checks pass.
9. Update the local main branch:
   ```bash
   git switch main
   git pull origin main
   ```

A Pull Request is a GitHub review and collaboration process, not a Git command.

## Question 10 – Difference between `git reset` and `git revert`. Give a use case for each and explain why revert is safer on a shared branch.

**Answer:**

**`git reset`**
- Moves the current branch pointer to a different commit.
- Depending on the option, it can also change the staging area and working directory.
- Commonly used to undo local commits that have not been shared.

Examples:

```bash
# Unstage changes while keeping working-directory files
git reset

# Move back one commit and keep changes staged
git reset --soft HEAD~1

# Move back one commit and keep changes unstaged in working directory
git reset --mixed HEAD~1
```

**`git revert`**
- Creates a new commit that reverses the changes introduced by an earlier commit.
- It does not remove the original commit from history.
- Commonly used to undo a commit that has already been pushed or shared.

Example:

```bash
git revert <commit-hash>
```

**When to use each:**
- Use `git reset` when a local commit was made by mistake and has not been shared, and you want to move the branch back or adjust staged changes.
- Use `git revert` when an incorrect commit has already been pushed to a shared branch and you need to undo its effect while preserving history.

**Why revert is safer on a shared branch:** Revert adds a new commit instead of rewriting or removing existing shared commits. Other developers can continue using the same history, reducing the risk of divergent histories and coordination problems.

**Caution:** `git reset --hard` discards uncommitted changes in the working directory and staging area. Use it only when you are certain those changes are no longer needed.
